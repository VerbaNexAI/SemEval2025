````python
import re
import json
from openai import OpenAI

def extract_json_from_text(text):
    """
    Improved JSON extraction with better error handling
    """
    try:
        # Look for complete JSON blocks with proper formatting
        json_match = re.search(r'{\s*"soft_labels":\s*\[.*?\]\s*}', text, re.DOTALL)
        if json_match:
            json_str = json_match.group()
            # Clean up any residual markdown formatting
            json_str = re.sub(r'```json\n?', '', json_str)
            return json.loads(json_str)
        return None
    except Exception as e:
        print(f"Error extracting JSON: {str(e)}")
        return None

def llm(query):
    print(query[0])
    # messages = [
    #     {"role": "system", "content": "Analyze text and call store_hallucination_analysis with results"},
    #     {"role": "user", "content": query[0]}
    # ]

    try:

        model_response = model_llm(query[0])
        result = extract_json_from_text(model_response)
        
        # Validate the result structure
        if result and "soft_labels" in result:
            for label in result["soft_labels"]:
                if not all(k in label for k in ["start", "end", "prob"]):
                    return {"soft_labels": []}
            return result
        return {"soft_labels": []}
        
    except Exception as e:
        print(f"Error: {str(e)}")
        return {"soft_labels": []}
````

````python
import json
from pathlib import Path

def validate_label_ranges(entry, text_length):
    """Validate and clean label ranges based on text length."""
    valid_soft = []
    valid_hard = []
    
    # Validate soft labels
    for label in entry.get('soft_labels', []):
        try:
            start = int(label.get('start', -1))
            end = int(label.get('end', -1))
            
            # Check if ranges are valid
            if (0 <= start < end <= text_length and 
                'prob' in label and 
                isinstance(label['prob'], (int, float)) and 
                0 <= label['prob'] <= 1):
                valid_soft.append({
                    'start': start,
                    'end': end,
                    'prob': float(label['prob'])
                })
        except (TypeError, ValueError):
            continue
    
    # Validate hard labels
    for label in entry.get('hard_label', []):
        try:
            start = int(label.get('start', -1))
            end = int(label.get('end', -1))
            
            # Check if ranges are valid
            if 0 <= start < end <= text_length:
                valid_hard.append({
                    'start': start,
                    'end': end
                })
        except (TypeError, ValueError):
            continue
    
    entry['soft_labels'] = valid_soft
    entry['hard_label'] = valid_hard
    return entry

def clean_jsonl_file():
    input_file = Path('responses.jsonl')
    output_file = Path('cleaned_responses.jsonl')
    
    with open(input_file, encoding='utf-8') as f_in, open(output_file, 'w', encoding='utf-8') as f_out:
        for line in f_in:
            try:
                entry = json.loads(line.strip())
                
                # Get text length for validation
                text_length = len(entry.get('model_output_text', ''))
                
                # Clean and validate the entry
                cleaned = validate_label_ranges(entry, text_length)
                
                # Write the cleaned entry
                f_out.write(json.dumps(cleaned, ensure_ascii=False) + '\n')
                
                # Print info for problematic entries (like tst-en-5)
                if entry['id'] == 'tst-en-5':
                    print(f"Cleaned entry {entry['id']}:")
                    print(f"Text length: {text_length}")
                    print(f"Original labels: {entry.get('soft_labels', [])}")
                    print(f"Cleaned labels: {cleaned['soft_labels']}")
                    
            except json.JSONDecodeError as e:
                print(f"Error processing line: {e}")
                continue

if __name__ == "__main__":
    clean_jsonl_file()
````

````python
# Import necessary libraries
import pandas as pd
from scipy.stats import spearmanr
import numpy as np
import json
import wandb
# evaluation = "deepseek/deepseek-r1-distill-llama-70b"
# Initialize W&B run
wandb.init(project="hallucination-detection", 
           job_type="evaluation",
           name=f"{evaluation}",
           tags=["baseline", evaluation])

def recompute_hard_labels(soft_labels):
    """Infer hard labels from the soft labels provided"""
    if not soft_labels:  # Handle empty soft_labels
        return []
        
    hard_labels = []
    prev_end = -1
    for start, end in (
        (lbl["start"], lbl["end"])
        for lbl in sorted(soft_labels, key=lambda span: (span["start"], span["end"]))
        if lbl["prob"] > 0.5
    ):
        if start == prev_end:
            hard_labels[-1][-1] = end
        else:
            hard_labels.append([start, end])
        prev_end = end
    return hard_labels

def load_jsonl_file_to_records(filename):
    """Read data from a JSONL file and format as records"""
    records = []
    with open(filename, 'r', encoding='utf-8') as f:
        for line in f:
            try:
                record = json.loads(line)
                # Ensure soft_labels exists
                if "soft_labels" not in record:
                    record["soft_labels"] = []
                # Compute hard_labels if not present
                if "hard_labels" not in record:
                    record["hard_labels"] = recompute_hard_labels(record["soft_labels"])
                # Add text_len for convenience
                record["text_len"] = len(record.get("model_output_text", ""))
                records.append(record)
            except json.JSONDecodeError:
                print(f"Error parsing line in {filename}")
                continue
    return records

def score_iou(ref_dict, pred_dict):
    """Compute IoU (Intersection over Union) between reference and prediction"""
    ref_spans = ref_dict["hard_labels"]
    pred_spans = pred_dict["hard_labels"]
    
    # Handle empty spans
    if not ref_spans and not pred_spans:
        return 1.0
    if not ref_spans or not pred_spans:
        return 0.0
    
    # Convert to sets of character positions for IoU calculation
    ref_chars = set()
    for start, end in ref_spans:
        ref_chars.update(range(start, end))
    
    pred_chars = set()
    for start, end in pred_spans:
        pred_chars.update(range(start, end))
    
    # Calculate IoU
    intersection = len(ref_chars.intersection(pred_chars))
    union = len(ref_chars.union(pred_chars))
    
    return intersection / union if union > 0 else 0.0

def score_cor(ref_dict, pred_dict):
    """Compute correlation between reference and prediction soft labels"""
    ref_soft = ref_dict.get("soft_labels", [])
    pred_soft = pred_dict.get("soft_labels", [])
    
    # Handle empty soft labels
    if not ref_soft or not pred_soft:
        return 0.0
    
    # Create character-level probability arrays
    text_len = max(ref_dict.get("text_len", 0), pred_dict.get("text_len", 0))
    ref_vec = [0.0] * text_len
    pred_vec = [0.0] * text_len
    
    # Fill in probabilities from soft labels
    for span in ref_soft:
        for idx in range(span["start"], span["end"]):
            if idx < text_len:
                ref_vec[idx] = span["prob"]
    
    for span in pred_soft:
        for idx in range(span["start"], span["end"]):
            if idx < text_len:
                pred_vec[idx] = span["prob"]
    
    # Check for constant values
    ref_unique = set(round(v, 8) for v in ref_vec)
    pred_unique = set(round(v, 8) for v in pred_vec)
    
    if len(ref_unique) == 1 or len(pred_unique) == 1:
        return float(len(ref_unique) == len(pred_unique))
    
    # Calculate Spearman correlation
    return spearmanr(ref_vec, pred_vec).correlation

# Main evaluation function
def evaluate_files(ref_file, pred_file, output_file=None):
    # Load data
    ref_dicts = load_jsonl_file_to_records(ref_file)
    pred_dicts = load_jsonl_file_to_records(pred_file)
    
    # Create ID mappings
    ref_id_map = {d["id"]: d for d in ref_dicts}
    pred_id_map = {d["id"]: d for d in pred_dicts}
    
    # Find common IDs
    common_ids = set(ref_id_map.keys()).intersection(set(pred_id_map.keys()))
    
    # Align data by ID
    aligned_refs = [ref_id_map[id] for id in sorted(common_ids)]
    aligned_preds = [pred_id_map[id] for id in sorted(common_ids)]
    
    # Calculate scores
    ious = np.array([score_iou(r, p) for r, p in zip(aligned_refs, aligned_preds)])
    cors = np.array([score_cor(r, p) for r, p in zip(aligned_refs, aligned_preds)])
    
    # Log metrics to W&B
    wandb.log({
        "IoU_mean": ious.mean(),
        "Cor_mean": cors.mean()
    })
    
    # Write to output file if specified
    if output_file:
        with open(output_file, "w") as f:
            f.write(f"IoU: {ious.mean():.8f}\n")
            f.write(f"Cor: {cors.mean():.8f}\n")
    
    return ious.mean(), cors.mean()

# Run the evaluation
ref_file = "mushroom.en-tst.v1.jsonl"
pred_file = "cleaned_responses.jsonl"
output_file = "scores.txt"

# Run evaluation and get metrics
iou_score, cor_score = evaluate_files(ref_file, pred_file, output_file)

# Display results
print(f"IoU Score: {iou_score:.8f}")
print(f"Correlation Score: {cor_score:.8f}")

# Finish the W&B run
wandb.finish()
````
